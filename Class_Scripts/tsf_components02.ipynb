```python
df = pd.read_csv('05_ComponentData.csv')
df['date'] = pd.to_datetime(df['date'])
df = df.set_index('date')
df
```

```python
plot_series(df['data'],
            x_label="Year", y_label="Earnings per share ($)",
            title="Quarterly earnings of Johnson & Johnson in million of USD")
plt.grid(color="grey", linestyle="--", linewidth=0.5)
plt.show()
```

```python
# Look at Documentation for seasonal_decompose and STL
# https://www.statsmodels.org/devel/generated/statsmodels.tsa.seasonal.seasonal_decompose.html
# https://www.statsmodels.org/devel/generated/statsmodels.tsa.seasonal.STL.html
# Reference: R. B. Cleveland, W. S. Cleveland, J.E. McRae, and I. Terpenning (1990) STL: A Seasonal-Trend Decomposition Procedure Based on LOESS. Journal of Official Statistics, 6, 3-73.
```

```python
advanced_decomposition = STL(df.data, period=4).fit()
```

```python
plot_series(advanced_decomposition.observed,
            x_label="Year", y_label="Observed data")
plt.grid(color="grey", linestyle="--", linewidth=0.5)
plt.show()                        
```

```python
plot_series(advanced_decomposition.trend,
            x_label="Year", y_label="Trend component")
plt.grid(color="grey", linestyle="--", linewidth=0.5)
plt.show()
```

```python
plot_series(advanced_decomposition.seasonal,
            x_label="Year", y_label="Seasonal component")
plt.grid(color="grey", linestyle="--", linewidth=0.5)
plt.show()
```

```python
plot_series(advanced_decomposition.resid,
            x_label="Year", y_label="Residual component")
plt.grid(color="grey", linestyle="--", linewidth=0.5)
plt.show()
```

```python
advanced_decomposition.plot()
plt.show()
```
