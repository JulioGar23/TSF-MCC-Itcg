```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from scipy.stats import boxcox
```

```python
df = pd.read_csv('09_airline_passengers.csv', index_col='Month', parse_dates=True)
df.head(10)
```

```python
ax = df['Passengers'].plot(
    figsize=(14, 5),
    color='royalblue',
    linewidth=2.5,
    title='Airline Passengers Over Time'
)

ax.set_xlabel('Time (years)', fontsize=14)
ax.set_ylabel('Number of Passengers', fontsize=14)

ax.grid(True, linestyle='--', alpha=0.4)
ax.tick_params(axis='both', labelsize=12)

plt.tight_layout()
plt.show()
```

```python
df['SqrtPassengers'] = np.sqrt(df['Passengers'])
ax = df['SqrtPassengers'].plot(
    figsize=(14, 5),
    color='royalblue',
    linewidth=2.5,
    title='Square Root'
)

ax.set_xlabel('Time (years)', fontsize=14)
ax.set_ylabel('Square Root of Number of Passengers', fontsize=14)

ax.grid(True, linestyle='--', alpha=0.4)
ax.tick_params(axis='both', labelsize=12)

plt.tight_layout()
plt.show()
```

```python
df['LogPassengers'] = np.log(df['Passengers'])
ax = df['LogPassengers'].plot(
    figsize=(14, 5),
    color='royalblue',
    linewidth=2.5,
    title='Log Transformation'
)

ax.set_xlabel('Time (years)', fontsize=14)
ax.set_ylabel('Log of Number of Passengers', fontsize=14)

ax.grid(True, linestyle='--', alpha=0.4)
ax.tick_params(axis='both', labelsize=12)

plt.tight_layout()
plt.show()
```

```python
data, lam = boxcox(df['Passengers'])
print(lam)
#print(data)
```

```python
df['BoxCoxPassengers'] = data
ax = df['BoxCoxPassengers'].plot(
    figsize=(14, 5),
    color='royalblue',
    linewidth=2.5,
    title='BOX-COX Transformation'
)

ax.set_xlabel('Time (years)', fontsize=14)
ax.set_ylabel('BOX-COX of Number of Passengers', fontsize=14)

ax.grid(True, linestyle='--', alpha=0.4)
ax.tick_params(axis='both', labelsize=12)

plt.tight_layout()
plt.show()
```
